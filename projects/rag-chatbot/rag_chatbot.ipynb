## Step 2: Import Libraries

Let's import all the tools we need:

```python
# Document loading and splitting
from langchain.document_loaders import TextLoader
from langchain.text_splitter import CharacterTextSplitter

# Embeddings and vector store
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS

# For creating sample documents
import os
```

✅ **Check:** If this cell runs without errors, all imports are working correctly!

## Step 3: Create Sample Documents

Let's create some sample knowledge for our chatbot to learn from:

```python
# Create a sample_docs directory
os.makedirs('sample_docs', exist_ok=True)

# Sample document about AI
sample_text = """
Artificial Intelligence (AI) is revolutionizing technology.
Machine Learning is a subset of AI that learns from data.
Deep Learning uses neural networks with many layers.
Natural Language Processing helps computers understand human language.
Computer Vision enables machines to interpret visual information.
Reinforcement Learning trains agents through rewards and penalties.
Generative AI can create new content like text, images, and music.
"""

# Save to file
with open('sample_docs/ai_basics.txt', 'w') as f:
    f.write(sample_text)

print("✅ Sample document created!")
print(f"📄 Content preview: {sample_text[:100]}...")
```

## Step 4: Load and Split Documents

Now let's load our document and split it into chunks:

```python
# Load the document
loader = TextLoader('sample_docs/ai_basics.txt')
documents = loader.load()

# Split into chunks (important for RAG!)
text_splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)
docs = text_splitter.split_documents(documents)

print(f"📚 Loaded {len(documents)} document(s)")
print(f"✂️ Split into {len(docs)} chunks")
print(f"\nFirst chunk: {docs[0].page_content}")
```

### 💡 Why Split Documents?
- Smaller chunks = more precise retrieval
- Overlap ensures context isn't lost
- Typical chunk size: 100-500 tokens

## Step 5: Create Embeddings

Embeddings convert text into numerical vectors that capture meaning:

```python
# Initialize the embedding model
embeddings = HuggingFaceEmbeddings(
    model_name='sentence-transformers/all-MiniLM-L6-v2'
)

print("✅ Embedding model loaded!")
print("💡 This model converts text into 384-dimensional vectors")
```

⏳ **Note:** First download may take 1-2 minutes. Subsequent runs will be faster!

## Step 6: Build Vector Store

The vector store allows fast similarity search:

```python
# Create the vector store
vectorstore = FAISS.from_documents(docs, embeddings)

print(f"✅ Vector store created with {vectorstore.index.ntotal} vectors!")
print("💡 FAISS enables fast similarity search")
```

## Step 7: Test Retrieval

Let's test if our system can find relevant information:

```python
# Try a search query
query = "What is machine learning?"
results = vectorstore.similarity_search(query, k=2)

print(f"🔍 Query: {query}\n")
for i, result in enumerate(results, 1):
    print(f"Result {i}: {result.page_content}\n")
```

## Step 8: Build a Simple Q&A Function

Let's create a reusable function for querying:

```python
def ask_rag(question, top_k=2):
    """
    Ask a question and retrieve relevant documents.
    
    Args:
        question: Your question string
        top_k: Number of results to return
    
    Returns:
        List of relevant text chunks
    """
    results = vectorstore.similarity_search(question, k=top_k)
    return [r.page_content for r in results]

# Test it!
answer = ask_rag("What are the types of AI?")
print("📝 Retrieved information:\n")
for i, chunk in enumerate(answer, 1):
    print(f"{i}. {chunk}\n")
```
